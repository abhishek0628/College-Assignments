```python
def LinearRgression(X,w,b):
    return np.dot(X,w)+b
```

```python
def cost_function(X,y,w,b):
    m=len(y)
    y_pred=LinearRegression(X,w,b)
    return (1/(2*m))*(np.sum((y-y_pred)**2))
```
